kappa4 looks like the right distribution to compare to the data. Maxwellian might also be a good fit.

```python
h, k = 0.1, 0
scale = 1
mean, var, skew, kurt = kappa4.stats(h, k, moments="mvsk")

print(f"Mean: {mean}, Variance: {var}, Skewness: {skew}, Kurtosis: {kurt}")

x = np.linspace(
    kappa4.ppf(0.01, h, k, scale=scale),
    kappa4.ppf(0.99, h, k, scale=scale),
    100,
)
plt.plot(
    x,
    kappa4.pdf(x, h, k, scale=scale, loc=0),
    "r-",
    lw=5,
    alpha=0.6,
    label="kappa4 pdf",
)
```
