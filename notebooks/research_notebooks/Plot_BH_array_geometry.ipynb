```python
import os

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
data_dir = os.path.join(os.path.dirname(""), os.pardir, os.pardir, "data")
coord_file = os.path.join(data_dir, "Surface_DAS_DTS_UTM_coordinates.csv")

df = pd.read_csv(coord_file)
df.head()
```

```python
N = 8672
ind = np.arange(53, N, 1)

x = [float(df.iloc[i]["X"]) for i in ind]
y = [float(df.iloc[i]["Y"]) for i in ind]
z = [float(df.iloc[i]["Z"]) for i in ind]
channel = np.array(df.iloc[ind]["Channel"])
n = len(x)
plt.scatter(x[:n], y[:n], c=ind)
plt.colorbar(label="Channel Number")
plt.xlabel("x UTM (m)", fontsize=12)
plt.locator_params(axis="x", nbins=5)

start_ch = 4500
nchannels = 1500
i = np.where(channel == str(start_ch))[0][0]
plt.plot(x[i : i + nchannels], y[i : i + nchannels], "r", linewidth=2)
```
